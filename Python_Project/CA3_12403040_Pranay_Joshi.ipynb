Q.1 Write code to plot a line chart using Pandas, showing the trend of a numerical
variable over time from a given dataset. Explain how to customize line color and
style in Pandas plots.

```python
import pandas as pd  
import matplotlib.pyplot as plt  
 # creatign a sample data 
data = {  
    'date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'value': [10, 15, 13, 17, 20, 18, 22, 24, 30, 28]  
}   
df = pd.DataFrame(data)  
df.set_index('date', inplace=True)  
plt.figure(figsize=(10, 5))  
plt.plot(df.index, df['value'], color='cyan', linestyle='-', linewidth=2, marker='o', markersize=5)  #giving the colors to the graphs 
plt.title('Trend of Value Over Time')  
plt.xlabel('Date')  
plt.ylabel('Value')  
plt.grid(True)   
plt.show()
```

Q2. Using Matplotlib, create a histogram to display the distribution of a continuous
variable. Describe how changing the number of bins affects the shape and
readability of the histogram.

```python
import numpy as np  
import matplotlib.pyplot as plt   
data = np.random.randn(1000) #generation random data    
def plot_histogram(data, bins):  
    plt.figure(figsize=(10, 6))  
    plt.hist(data, bins=bins, edgecolor='black', alpha=0.7)  
    plt.title(f"Histogram with {bins} Bins")  
    plt.xlabel("Value")  
    plt.ylabel("Frequency")  
    plt.grid(axis='y', alpha=0.75)  
    plt.show()   
for num_bins in [20, 50]:  # this is the code where the no. of bin is changing using loop 
    plot_histogram(data, num_bins)
```

Q5. Describe the components of a box and whisker plot. Using Matplotlib, plot a box
plot for a dataset with multiple groups and analyze the spread and outliers of each
group.

```python
import numpy as np  
import matplotlib.pyplot as plt  

# Simulating dataset for three groups  
np.random.seed(42)  
group1 = np.random.normal(loc=50, scale=10, size=100)  # Group 1  
group2 = np.random.normal(loc=60, scale=15, size=100)  # Group 2  
group3 = np.random.normal(loc=70, scale=20, size=100)  # Group 3  

data = [group1, group2, group3]  

# Creating a box plot  
plt.figure(figsize=(10, 6))  
plt.boxplot(data, labels=['Group 1', 'Group 2', 'Group 3'])  

# Adding titles and labels  
plt.title('Box Plot of Multiple Groups')  
plt.ylabel('Values')  
plt.grid(axis='y')  
plt.show()
```

Q6.Explain what a heatmap represents and use Seaborn to create a heatmap from a
correlation matrix of a dataset. Describe how color gradients in the heatmap can help
interpret the data.

```python
import pandas as pd  
import numpy as np  
import matplotlib.pyplot as plt  
import seaborn as sns  

df = pd.read_csv("C:\\Users\\Lenovo\\Desktop\\dataset\\iris_data.csv")  
print("Data types of the columns:")  
print(df.dtypes)  

if 'species' in df.columns:  
    df_numeric = df.drop(columns=['species'])  
else:  
    df_numeric = df.select_dtypes(include=[np.number])  

correlation_matrix = df_numeric.corr()  

plt.figure(figsize=(10, 8))  
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", square=True, cbar_kws={"shrink": .8})  
plt.title('Correlation Matrix of Iris Dataset')  
plt.show()
```
